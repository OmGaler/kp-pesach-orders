# Sales analysis

Refresh the workbook export, then inspect category share, product-type rollups and a few lightweight charts locally.

```python
from pathlib import Path
import os
import subprocess

candidate_roots = [Path.cwd(), Path.cwd().parent]
project_root = next(
    (path for path in candidate_roots if (path / 'package.json').exists()),
    Path.cwd()
)

npm_command = 'npm.cmd' if os.name == 'nt' else 'npm'
subprocess.run([npm_command, 'run', 'analyse:sales'], cwd=project_root, check=True)
```

```python
import json

import matplotlib.pyplot as plt
import pandas as pd

data_dir = project_root / 'notebooks' / 'data'
analysis = json.loads((data_dir / 'sales-analysis.json').read_text())
category_totals = pd.read_csv(data_dir / 'category-totals.csv')
subcategory_totals = pd.read_csv(data_dir / 'subcategory-totals.csv')
product_type_totals = pd.read_csv(data_dir / 'product-type-totals.csv')
product_totals = pd.read_csv(data_dir / 'product-totals.csv')
product_type_size_mix = pd.read_csv(data_dir / 'product-type-size-mix.csv')
delivery_date_totals = pd.read_csv(data_dir / 'delivery-date-totals.csv')
analysis['summary']
```

```python
category_totals[['category', 'totalQuantity', 'quantityPct']]
```

```python
plt.figure(figsize=(10, 5))
chart = category_totals.head(25).sort_values('totalQuantity')
plt.barh(chart['category'], chart['totalQuantity'], color='#3f7d58')
plt.title('Top categories by quantity sold')
plt.xlabel('Quantity sold')
plt.tight_layout()
plt.show()
```

```python
subcategory_totals[['category', 'subcategory', 'totalQuantity', 'categoryQuantityPct']].head(25)
```

```python
plt.figure(figsize=(10, 6))
chart = product_type_totals.head(25).sort_values('totalQuantity')
labels = chart['productType'] + ' (' + chart['category'] + ')'
plt.barh(labels, chart['totalQuantity'], color='#7aa874')
plt.title('Top product types')
plt.xlabel('Quantity sold')
plt.tight_layout()
plt.show()
```
